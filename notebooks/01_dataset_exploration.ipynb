```python
import json
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

metadata_path = (
    PROJECT_ROOT
    / "data/raw/Sen1Floods11/v1.1/Sen1Floods11_Metadata.geojson"
)

with open(metadata_path, "r", encoding="utf-8") as file:
    metadata = json.load(file)

events = pd.json_normalize(metadata["features"])

print("Number of event records:", len(events))
print("\nAvailable columns:")
print(events.columns.tolist())

print("\nEvent summary:")

display(
    events[
        [
            "properties.location",
            "properties.s1_date",
            "properties.orbit",
            "properties.train_chip",
            "properties.val_chip",
        ]
    ]
)
```
